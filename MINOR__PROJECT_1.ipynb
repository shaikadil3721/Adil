### `lib/screens/quiz_screen.dart`  
This contains the core logic using `setState()` to handle user interaction.

```python
print('''
import 'package:flutter/material.dart';
import '../models/question.dart';
import '../data/questions_data.dart';
import 'results_screen.dart';

class QuizScreen extends StatefulWidget {
  const QuizScreen({super.key});

  @override
  State<QuizScreen> createState() => _QuizScreenState();
}

class _QuizScreenState extends State<QuizScreen> {
  int _questionIndex = 0;
  int _totalScore = 0;

  void _answerQuestion(int selectedIndex) {
    if (selectedIndex == quizQuestions[_questionIndex].correctAnswerIndex) {
      _totalScore++;
    }

    setState(() {
      _questionIndex++;
    });

    if (_questionIndex >= quizQuestions.length) {
      Navigator.of(context).pushReplacement(
        MaterialPageRoute(
          builder: (context) => ResultsScreen(
            score: _totalScore,
            totalQuestions: quizQuestions.length,
          ),
        ),
      );
    }
  }

  @override
  Widget build(BuildContext context) {
    return Scaffold(
      appBar: AppBar(title: Text('Question ${_questionIndex + 1}')),
      body: _questionIndex < quizQuestions.length
          ? Column(
              children: [
                Text(quizQuestions[_questionIndex].questionText),
                ...quizQuestions[_questionIndex].options.asMap().entries.map((entry) {
                  return ElevatedButton(
                    onPressed: () => _answerQuestion(entry.key),
                    child: Text(entry.value),
                  );
                }).toList(),
              ],
            )
          : const Center(child: CircularProgressIndicator()),
    );
  }
}
''')
```

### `lib/screens/results_screen.dart`  
Displays the final score and results.

```python
print('''
import 'package:flutter/material.dart';

class ResultsScreen extends StatelessWidget {
  final int score;
  final int totalQuestions;

  const ResultsScreen({super.key, required this.score, required this.totalQuestions});

  @override
  Widget build(BuildContext context) {
    return Scaffold(
      body: Center(
        child: Column(
          mainAxisAlignment: MainAxisAlignment.center,
          children: [
            Text('Quiz Finished!', style: TextStyle(fontSize: 24)),
            Text('Your Score: $score / $totalQuestions'),
            ElevatedButton(
              onPressed: () => Navigator.of(context).popToRoot(),
              child: const Text('Restart Quiz'),
            ),
          ],
        ),
      ),
    );
  }
}
''')
```
